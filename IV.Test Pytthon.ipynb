```python
# CÂU 1: Tìm cặp số đầu tiên có tổng = target
def find_pair(nums, target):
    seen = {}
    for num in nums:
        complement = target - num
        if complement in seen:
            return (complement, num)
        seen[num] = True
    return None


# Test
nums = [2, 7, 11, 15, 3, 6]
target = 26
print(find_pair(nums, target))   
```

```python
# CÂU 2: Tìm dãy con liên tiếp dài nhất có tổng chia hết cho k
def find_longest_subarray(nums, k):
    mod_map = {0: -1}  # remainder 0 xuất hiện trước index 0

    prefix_sum = 0
    best_start = -1
    best_len = 0

    for i in range(len(nums)):
        prefix_sum += nums[i]
        remainder = prefix_sum % k

        if remainder in mod_map:
            length = i - mod_map[remainder]
            if length > best_len:
                best_len = length
                best_start = mod_map[remainder] + 1
        else:
            mod_map[remainder] = i

    if best_len == 0:
        return None

    return nums[best_start : best_start + best_len]


# Test
nums = [2, 7, 6, 1, 4, 5]
k = 3
print(find_longest_subarray(nums, k))   # [7, 6, 1, 4]

nums2 = [1, 2, 3]
k2 = 7
print(find_longest_subarray(nums2, k2)) # None
```
